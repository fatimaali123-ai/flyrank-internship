## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(df.shape)

print("\nCTR Buckets")

df["ctr_bucket"] = pd.cut(
    df["ctr"],
    bins=[-1,1,3,100],
    labels=["Low","Medium","High"]
)

ctr_table = df.groupby("ctr_bucket").agg(
    n=("content_id","count"),
    avg_impressions=("impressions_90d","mean"),
    avg_clicks=("clicks_90d","mean")
)

print(ctr_table)

print("\nVerdict: CONFIRMED")


print("\nImpression Buckets")

df["imp_bucket"] = pd.qcut(
    df["impressions_90d"],
    q=3,
    labels=["Low","Medium","High"]
)

imp_table = df.groupby("imp_bucket").agg(
    n=("content_id","count"),
    avg_ctr=("ctr","mean")
)

print(imp_table)

print("\nVerdict: MIXED")
```

## My Rule

I will prioritize content that has high impressions but a low CTR because these pages are already visible in search results and improving their click-through rate may increase traffic.

Signals used:
1. CTR
2. Impressions

Reason Code:
LOW_CTR_HIGH_IMPRESSIONS

Action:
Improve CTR (title, meta description, and snippet optimization)

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import os

df["baseline_score"] = (
    df["impressions_90d"] * (1 - df["ctr"]/100)
)

df["reason_code"] = "LOW_CTR_HIGH_IMPRESSIONS"

df["action"] = "Improve CTR"

ranked = df.sort_values(
    "baseline_score",
    ascending=False
)

os.makedirs("work/outputs", exist_ok=True)

ranked.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print(ranked[[
    "content_id",
    "baseline_score",
    "reason_code",
    "action"
]].head())
```

The baseline score prioritizes pages that already receive many impressions but have a low CTR.

Reason Code:
LOW_CTR_HIGH_IMPRESSIONS

Action:
Improve CTR

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

```python
top20 = ranked.head(20)

for i,row in top20.iterrows():

    print("="*60)

    print("Content:",row["content_id"])

    print("Action:",row["action"])

    print("Reason:",row["reason_code"])

    print("Confidence: Medium")

    print("What would make it wrong?")

    print("- Seasonal traffic")
    print("- Already optimized metadata")
    print("- Brand queries dominating clicks")
```
