```python
import pandas as pd
import numpy as np

from sklearn.model_selection import GroupShuffleSplit
from sklearn.ensemble import RandomForestClassifier

# =========================================================
# 1. Load the dataset
# =========================================================

DATA_PATH = "/content/flyrank-internship/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)


# =========================================================
# 2. Create the declining target
# =========================================================

df["is_declining_label"] = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .astype(int)
)

print("\nTarget distribution:")
print(df["is_declining_label"].value_counts())

print("\nDeclining rate:")
print(df["is_declining_label"].mean())


# =========================================================
# 3. Honest grouped train/test split
# =========================================================

RANDOM_STATE = 42

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=RANDOM_STATE
)

train_idx, test_idx = next(
    gss.split(
        df,
        y=df["is_declining_label"],
        groups=df["client_id"]
    )
)

train = df.iloc[train_idx].copy()
test = df.iloc[test_idx].copy()

print("\nTrain shape:", train.shape)
print("Test shape:", test.shape)

print("\nTrain clients:", train["client_id"].nunique())
print("Test clients:", test["client_id"].nunique())

overlap = set(train["client_id"]).intersection(
    set(test["client_id"])
)

print("\nClient overlap:", overlap)


# =========================================================
# 4. Feature selection
# =========================================================

excluded = [
    "content_id",
    "client_id",
    "is_declining_label",
    "trend_direction",
    "trend_pct"
]

feature_cols = [
    col for col in train.columns
    if col not in excluded
    and pd.api.types.is_numeric_dtype(train[col])
]

print("\nNumber of features:", len(feature_cols))


# =========================================================
# 5. X and y
# =========================================================

X_train = train[feature_cols].copy()
X_test = test[feature_cols].copy()

y_train = train["is_declining_label"].astype(int)
y_test = test["is_declining_label"].astype(int)


# =========================================================
# 6. Missing values
# =========================================================

train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)


# =========================================================
# 7. Train Random Forest
# =========================================================

model_honest = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=10,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

model_honest.fit(X_train, y_train)

honest_scores = model_honest.predict_proba(X_test)[:, 1]

print("\nHonest grouped model trained successfully.")


# =========================================================
# 8. Precision@K
# =========================================================

def precision_at_k(y_true, scores, k):
    y_true = np.asarray(y_true)
    scores = np.asarray(scores)

    top_indices = np.argsort(-scores)[:k]

    return y_true[top_indices].mean()


honest_p20 = precision_at_k(
    y_test,
    honest_scores,
    20
)

honest_p50 = precision_at_k(
    y_test,
    honest_scores,
    50
)

print("\nHonest grouped model:")
print("Precision@20:", honest_p20)
print("Precision@50:", honest_p50)


# =========================================================
# 9. Comparison with Week 5
# =========================================================

comparison = pd.DataFrame({
    "Evaluation": [
        "Week-5 model",
        "Week-6 honest grouped model"
    ],
    "Precision@20": [
        1.00,
        honest_p20
    ],
    "Precision@50": [
        1.00,
        honest_p50
    ]
})

print("\nModel comparison:")
display(comparison)
```

## 1. Ranked actions + reason codes

### Ranked actions

The validated grouped model is used to rank content items by their estimated likelihood of belonging to the declining-content class.

The ranked queue is a prioritization tool for human review. Higher-ranked items should be reviewed first, but the model score does not prove that a content refresh will improve performance.

### Reason codes

- `REFRESH_DECLINING`: the item is showing a declining trend and is therefore a candidate for human review for a possible refresh.
- `REVIEW_HIGH_SCORE`: the model ranks the item highly, but the underlying evidence should be inspected before choosing an action.
- `REVIEW_RECENT_UPDATE`: the content has a recent update signal, so the reviewer should avoid assuming that another refresh is immediately needed.
- `REVIEW_ONLY`: available evidence is insufficient to recommend a specific action.

### Archetype → action mapping

| Content archetype | Human-review action |
|---|---|
| Declining content | Review for a targeted content refresh |
| High model score without clear decline evidence | Inspect supporting metrics before deciding on an action |
| Recently updated content | Check whether enough time has passed to observe the effect of the update |
| Older content with declining performance | Review relevance, search intent, and content freshness |
| Insufficient evidence | Human review only; do not prescribe an automatic change |

### Decay / refresh insight

The dataset contains content-age and freshness signals, including `content_age_days`, `days_since_last_update`, `age_tier`, and `freshness_tier`, as well as performance-trend fields such as `trend_direction` and `trend_pct`.

These fields allow content age and freshness to be examined alongside performance decline. However, the presence of these variables does not by itself establish that content age or time since update causes decline.

For this playbook, the practical interpretation is therefore directional: older or less recently updated content that is also showing a declining performance signal may be worth prioritizing for human review. The recommendation is to inspect the content before deciding whether a refresh is appropriate.

### Cost / value thinking

The queue is most useful when it helps allocate limited analyst time to items where review is likely to be valuable.

Higher-priority review is more appropriate when the item has a meaningful performance signal, sufficient supporting data, and a potentially actionable issue.

Lower-value cases include items with little evidence, missing important inputs, or recent updates that have not had enough time to produce a measurable signal.

The model therefore helps prioritize human attention; it does not determine the economic value of a content change by itself.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

### Intended use

This playbook is intended to help a human analyst prioritize which content items should be reviewed first.

The model is used as decision-support and triage. A higher model score means that an item is ranked higher for review based on the evaluated model, but it does not prove that the content is declining because of a specific cause or that a refresh will improve performance.

The intended workflow is:

1. Generate the ranked queue.
2. Review the highest-priority items first.
3. Inspect the reason code and supporting metrics.
4. Decide whether a content action is appropriate.
5. Record the human decision.
6. Measure the outcome later where possible.

### Limits

The model was evaluated using a client-grouped train/test split. The validation result therefore describes performance on the evaluated test split, not guaranteed future performance.

The model should not be interpreted as a causal model. A high score does not prove that refreshing a page will increase traffic, clicks, CTR, or rankings.

Performance may also change when the data distribution changes, when content types differ from the evaluated data, when important input fields are missing, or when the business objective changes.

The queue is therefore a research and planning artifact for human review, not a production automation system.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human review rules

Every model-ranked item must be reviewed by a human before any content or SEO action is taken.

The reviewer should check:

1. Whether the content is still relevant.
2. Whether the observed performance issue is meaningful.
3. Whether the reason code matches the underlying metrics.
4. Whether the content was recently updated.
5. Whether there are editorial, business, legal, or brand constraints.
6. Whether the proposed action is appropriate and reversible.
7. Whether the available evidence is sufficient to justify spending review time on the item.

### No-go list

The following actions should NOT be automated:

- Automatically publishing rewritten content.
- Automatically deleting content.
- Automatically changing URLs or redirects.
- Automatically changing canonical tags or technical SEO settings.
- Automatically making claims about search-engine ranking causes.
- Automatically changing content when important evidence is missing.
- Automatically applying the same action to every content type.
- Automatically treating a high model score as proof that a refresh will improve performance.

The model should only prioritize items for review. Final content and SEO decisions remain with a human reviewer.

```python
# =========================================================
# Section 3 — Human review policy
# =========================================================

review_required = True

no_go_actions = [
    "Automatic publishing",
    "Automatic deletion",
    "Automatic URL/redirect changes",
    "Automatic canonical/technical SEO changes",
    "Automatic causal claims",
    "Automatic action when evidence is missing",
    "Automatic application across all content types"
]

print("Human review required:", review_required)

print("\nNo-go actions:")
for action in no_go_actions:
    print("-", action)

assert review_required is True

print("\nHuman-review policy check: PASS")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

### Monitoring

The playbook should be monitored over time rather than assumed to remain valid indefinitely.

Useful monitoring signals include:

- Changes in the distribution of model scores.
- Changes in the number of items entering the high-priority queue.
- Changes in input-feature availability or missingness.
- Changes in the types of content represented in the queue.
- Decline in Precision@20 or Precision@50 on a later evaluation period.
- Evidence that human reviewers find the queue less useful.
- Major changes in the underlying search or analytics data.

### Retrain / revalidate triggers

The model should be re-evaluated when:

1. Precision@20 or Precision@50 falls materially below the validated result.
2. Input distributions change substantially.
3. Important features become unavailable or change definition.
4. The content population changes substantially.
5. The evaluation period becomes substantially different from the original validation period.
6. A new label definition or business objective is introduced.

Retraining should not be triggered because of one or two incorrect recommendations. The decision should be based on systematic evidence that model performance or data validity has changed.

This is a lightweight research monitoring plan, not a production MLOps system.

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

### Ranked action queue

The validated model score is used to rank content items for human review.

The queue is a prioritization artifact. It does not represent an automatic decision to refresh, rewrite, delete, or publish content.

The exported queue contains the ranking, model score, reason code, and recommended human-review action.

The queue is exported to `work/outputs/` so that the research paper can reuse the same artifact without manually rebuilding it.

Generated queue data is kept out of git according to the repository workflow. The notebook should regenerate the queue when it is executed.

```python
# =========================================================
# Section 5 — Build ranked action queue
# =========================================================

from pathlib import Path

# Project directories
project_root = Path("/content/flyrank-internship")

outputs_dir = project_root / "work" / "outputs"
outputs_dir.mkdir(parents=True, exist_ok=True)

# =========================================================
# Create ranked queue from the honest test predictions
# =========================================================

ranked_queue = test[
    [
        "content_id",
        "client_id",
        "trend_direction",
        "trend_pct"
    ]
].copy()

# Add model score
ranked_queue["model_score"] = honest_scores

# Add actual label for research/audit purposes
ranked_queue["actual_declining_label"] = y_test.to_numpy()

# Highest score = highest review priority
ranked_queue = ranked_queue.sort_values(
    "model_score",
    ascending=False
).reset_index(drop=True)

# Rank
ranked_queue["rank"] = np.arange(
    1,
    len(ranked_queue) + 1
)

# =========================================================
# Reason codes
# =========================================================

def get_reason_code(row):
    trend = str(row["trend_direction"]).lower()
    trend_pct = row["trend_pct"]

    if trend == "down":
        return "REFRESH_DECLINING"

    if pd.notna(trend_pct) and trend_pct < 0:
        return "REFRESH_DECLINING"

    return "REVIEW_ONLY"


ranked_queue["reason_code"] = ranked_queue.apply(
    get_reason_code,
    axis=1
)

# =========================================================
# Recommended human-review action
# =========================================================

def get_action(reason_code):
    if reason_code == "REFRESH_DECLINING":
        return "Review content for a possible targeted refresh"

    return "Human review before deciding on any content action"


ranked_queue["recommended_action"] = (
    ranked_queue["reason_code"].apply(get_action)
)

# =========================================================
# Arrange columns
# =========================================================

ranked_queue = ranked_queue[
    [
        "rank",
        "content_id",
        "client_id",
        "model_score",
        "reason_code",
        "recommended_action",
        "trend_direction",
        "trend_pct",
        "actual_declining_label"
    ]
]

# =========================================================
# Export
# =========================================================

queue_path = outputs_dir / "w07_ranked_action_queue.csv"

ranked_queue.to_csv(
    queue_path,
    index=False
)

print("Queue exported successfully.")
print("Path:", queue_path)
print("Rows:", len(ranked_queue))

display(ranked_queue.head(20))
```
