## Abstract

This study examines whether historical search, traffic, engagement, content-age, and position signals can help prioritize content items for human review when they show signals associated with decline. The analysis addresses a practical content-performance problem: SEO and content teams may have many pages to review but limited time and therefore need a systematic way to decide where to look first. Using pseudonymized content data, I developed and evaluated a ranking workflow that produces a review-priority score, reason code, and recommended human-review action. The model is intended as decision support rather than an automatic content-refresh decision, with the final action remaining with a human analyst.


## Introduction

Content teams and SEO analysts often need to decide which pages deserve attention first. When many content items are available for review, manually examining every page with the same level of priority is inefficient. This creates a practical prioritization problem: can available historical performance signals help identify which content items should receive human review earlier?

This project addresses that problem in the context of FlyRank content performance analysis. The goal is not to automatically decide that a page is declining or that it must be refreshed. Instead, the goal is to build a repeatable ranking workflow that helps an editor or SEO analyst prioritize content items for investigation.

The research question is whether historical search, traffic, engagement, content-age, and position signals can be used to rank content items for human review when they show signs associated with decline. The resulting output is a review-priority score with a reason code and recommended human-review action. This framing keeps the model's role focused on decision support while leaving the final content decision to a human reviewer.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

### Data

This analysis uses the FlyRank ML Internship starter content-refresh dataset containing 30,000 pseudonymized content items and 44 available columns.

The target is `is_declining_label`, defined as 1 when `trend_direction` is `"down"` and 0 otherwise.

For modeling, identifier and outcome-derived fields were excluded. In particular, `content_id` and `client_id` were not used as predictive features, while `trend_direction` and `trend_pct` were excluded to avoid target leakage.

The model uses 29 numeric features covering search demand, competition, content size, historical impressions and clicks, recent activity, content age, freshness, CTR, position, engagement, scrolling, and AI-traffic signals.

No client names, domains, private queries, credentials, or other client-identifying information are included in the public analysis.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### Methodology

I use a Random Forest classifier to produce a decline-risk score for content prioritization.

The model uses 29 numeric features. Missing numeric values are filled using medians calculated from the training set only.

The target is `is_declining_label`, where 1 represents content whose observed `trend_direction` is `"down"`.

To reduce leakage across related observations, validation uses a client-grouped train/test split. The test set contains clients that do not appear in the training set.

The baseline is the existing transparent Week-4 hand-written score based on impressions and CTR. The Random Forest and baseline are evaluated on the same held-out test set using Precision@20 and Precision@50.

Identifier fields and label-derived fields are excluded from model features. The results are interpreted as measured performance on the evaluated split, not as proof of future or causal performance.

## 5. Limitations

*What this work cannot claim.*

### Limitations

This analysis has several important limitations.

First, the evaluation describes performance on the available client-grouped test split and does not guarantee performance on future data.

Second, the model is not causal. A high model score does not prove that refreshing a page will increase traffic, clicks, CTR, or rankings.

Third, model performance may change when the data distribution, content mix, available fields, or business objective changes.

Finally, the ranked queue is a human-review prioritization tool rather than an automated publishing, deletion, SEO, or refresh system.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

### Ranked recommendations

The model output should be used as a review queue.

#### Priority 1 — Review high-score declining content

Items with high model scores and an observed declining trend should receive the earliest human review.

**Reason code:** `REFRESH_DECLINING`

**Recommended action:** Review the content for a possible targeted refresh.

#### Priority 2 — Review high-score items without sufficient evidence

High model scores should not automatically trigger an edit. A reviewer should verify that the underlying performance signal is meaningful and that the content is still relevant.

**Reason code:** `REVIEW_ONLY`

**Recommended action:** Human review before deciding on any content action.

#### Priority 3 — Do not automate content changes

The model should never automatically publish rewrites, delete content, change URLs, modify technical SEO settings, or treat a high score as proof that a refresh will improve performance.

The final decision remains with a human reviewer.
